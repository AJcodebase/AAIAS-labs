```python
# Greedy best fit scearch
from queue import PriorityQueue
 
graph = {
    'a': [('b', 9), ('c', 4), ('d', 7)],
    'b': [('e', 11)],
    'c': [('e', 17), ('f', 12)],
    'd': [('f', 14)],
    'e': [('z', 5)],
    'f': [('z', 9)],
    'z': []
}
 
heuristic = {
    'a': 21,
    'b': 14,
    'c': 18,
    'd': 18,
    'e': 5,
    'f': 8,
    'z': 0
}
open_list = [('a',['a'],0)]
visited =[]
goal = 'z'
 
while open_list:
        best = 0
        for i in range(len(open_list)):
            if heuristic[open_list[i][0]] < heuristic[open_list[best][0]]:
                best = 1
        current, path, cost = open_list.pop(best)

        if current in visited:
            continue
        visited.append(current)

        print("Visited: ", current)
        print("Path: ", path)
        print("Cost: ", cost)

        if current == goal:
            print("Goal Found")
            print("Final Path: ","->".join(path))
            print("Total cost:" , cost)
            break
        for neighbor,edge_cost in graph[current]:
            if neighbor not in visited:
                open_list.append(
                (
                    neighbor,
                    path+[neighbor],
                    cost+edge_cost
                )
            )
```
