```python
from collections import deque
```

```python
graph = {
    'A' : ['B','C'],
    'B' : ['D','E'],
    'C' : ['F','G'],
    'D' : [],
    'E' : [],
    'F' : [],
    'G' : [],
}


visited = set()
goal = 'E'
queue = deque()
queue.append('A')
visited.add('A')

print("BFS Travel: ", end="")
while queue:
    node = queue.popleft()
    print(node, end=" ")
    
    if node == goal: 
        break
        
    for neighbour in graph[node]:
        if neighbour not in visited:
            visited.add(neighbour)
            queue.append(neighbour)
```

```python
graph = {
    'A' : ['B','C'],
    'B' : ['D','E'],
    'C' : ['F','G'],
    'D' : [],
    'E' : [],
    'F' : [],
    'G' : [],
}

parent = {'A': None} 

visited = set()
goal = 'E'
queue = deque()
queue.append('A')
visited.add('A')

print("BFS Travel: ", end="")
while queue:
    node = queue.popleft()
    print(node, end=" ")
    
    if node == goal: 
        break
        
    for neighbour in graph[node]:
        if neighbour not in visited:
            visited.add(neighbour)
            parent[neighbour] = node 
            queue.append(neighbour)

# Reconstruct the path
path = []
current = goal

while current is not None:
    path.append(current)
    current = parent[current]
path.reverse()

print("shortest Path:", " -> ".join(path))
```

```python
graph = {
    'a': ['b', 'c', 'd'],
    'b': ['a', 'e'],
    'c': ['a', 'e', 'f'],
    'd': ['a', 'f'],
    'e': ['b', 'c', 'z'],
    'f': ['c', 'd', 'z'],
    'z': ['e', 'f']
}
 
start = 'a'
goal = 'z'
 
def bfs_shortest_path(graph, start, goal):
    visited = set()
    queue = deque()
    queue.append(start)
    visited.add(start)
    parent = {start: None}
 
    print("BFS Traversal:", end=" ")
    while queue:
        node = queue.popleft()
        print(node, end=" ")
 
        if node == goal:
            break
 
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = node
                queue.append(neighbor)
    print()
 
    if goal in parent:
        path = []
        current = goal
        while current is not None:
            path.append(current)
            current = parent[current]
        path.reverse()
        print("BFS Path:", " -> ".join(path))
    else:
        print(f"No path found from {start} to {goal}")
 

bfs_shortest_path(graph, start, goal)
print()
```

```python
def dfs_path(graph, start, goal):
    visited = set()
    stack = [start]
    visited.add(start)
    parent = {start: None}
 
    print("DFS Traversal:", end=" ")
    while stack:
        node = stack.pop()
        print(node, end=" ")
 
        if node == goal:
            break
 
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = node
                stack.append(neighbor)
    print()
 
    if goal in parent:
        path = []
        current = goal
        while current is not None:
            path.append(current)
            current = parent[current]
        path.reverse()
        print("DFS Path:", " -> ".join(path))
    else:
        print(f"No path found from {start} to {goal}")
dfs_path(graph, start, goal)
```
