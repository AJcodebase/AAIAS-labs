```python
memory = {"A": "Unknown", "B": "Unknown"}

environment = [
    ("A", "Dirty"),
    ("B", "Dirty"),       # ENvironment fixed
    ("A", "Clean"),
    ("B", "Clean")
]

# initial agent position

agent_loc = "A"

for room, status in environment:
    memory[room] = status

    if status == "Dirty":
        action = "Suck"        # action when room is dirty, no action when clean
        memory[room] = "Clean"
    elif room == "A":
        action = "Move Right"
        agent_loc = "B"        # movement after cleaning to B
    else:
        action = "Move Left"
        agent_loc = "A"       # movment after cleaning to A

    print(f"Room: {room}, Status: {status}")
    print("Action:", action)
    print("Memory:", memory)
    print("Current location: ",agent_loc)
    print()
```

```python
memory = {"A": "Unknown", "B": "Unknown","C": "Unknown","D": "Unknown"}

environment = [
    ("A", "Dirty"),
    ("B", "Dirty"),
    ("A", "Clean"),
    ("B", "Clean"),
    ("C", "Dirty"),
    ("C", "Clean"),
    ("D", "Dirty"),
    ("D", "Clean")
]

agent_loc = "A"

for room, status in environment:
    memory[room] = status

    if status == "Dirty":
        action = "Suck"        # action when room is dirty, no action when clean
        memory[room] = "Clean"
    elif room == "A":
        action = "Move Right"
        agent_loc = "B"
    elif room == "B":
        action = "Move Right"
        agent_loc = "C"
    elif room == "C":
        action = "Move Right"
        agent_loc = "D"# movement after cleaning to B
    else:
        action = "Move Left"
        agent_loc = "A"       # movment after cleaning to A

    print(f"Room: {room}, Status: {status}")
    print("Action:", action)
    print("Memory:", memory)
    print("Current location: ",agent_loc)
    print()
```
